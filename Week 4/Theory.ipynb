```python
import numpy as np
from case_studies import *

def wolfe_search(f, grad_f, xk, pk, alpha_init=1.0, c1=1e-4, c2=0.9):
    def g(alpha):
        return f(xk + alpha * pk)
    
    def g_prime(alpha):
        return np.dot(grad_f(xk + alpha * pk), pk) # g'(alpha)
    
    l, u = 0, alpha_init
    
    while True:
        if g(u) > g(0) + c1 * u * g_prime(0) or g(u) > g(l):
            break
        if abs(g_prime(u)) < c2 * abs(g_prime(0)):
            return u
        if g_prime(u) > 0:
            break
        else:
            u *= 2
    
    while True:
        a = (l + u) / 2
        if g(a) > g(0) + c1 * a * g_prime(0) or g(a) > g(l):
            u = a
        else:
            if abs(g_prime(a)) < c2 * abs(g_prime(0)):
                return a
            if g_prime(a) < 0:
                l = a
            else:
                u = a
```

```python
# Define the parameters for the search
c1 = 0.1
c2 = 0.9
#x_k = 0 + 2.5 * np.random.randn(2)
x_k = np.array([3.46, -1.06])
p_k = -df3(x_k)  # Use the negative gradient as the search direction
alpha_init = 1.0

# Perform the Wolfe search
alpha_star = wolfe_search(f3, df3, x_k, p_k, alpha_init, c1, c2)
x_k1 = x_k + alpha_star * p_k

# Output the results
print("Starting point x_k:", x_k)
print("Search direction p_k:", p_k)
print("Step size alpha_star:", alpha_star)
print("New point x_k+1:", x_k1)

print(f"f3(x_k) = {f3(x_k)}")
print(f"f3(x_k1) = {f3(x_k1)}")
print(f"df3(x_k) = {df3(x_k)}")
print(f"df3(x_k1) = {df3(x_k1)}")
print(f"pkT * df(x_k) = {np.dot(-df3(x_k), df3(x_k))}")
print(f"pk1T * df(x_k1) = {np.dot(-df3(x_k1), df3(x_k1))}")

# Verifying Wolfe conditions
s_k = alpha_star * p_k
y_k = df3(x_k1) - df3(x_k)
sTy = np.dot(s_k, y_k)
yTy = np.dot(y_k, y_k)
sTs = np.dot(s_k, s_k)

print(f"sk = {s_k}")
print(f"yk = {y_k}")

print("s^T y:", sTy)
print("y^T y:", yTy)
print("s^T s:", sTs)
print("Conditions satisfied:", sTy < yTy and sTy < sTs)
```
