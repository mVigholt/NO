```python
import numpy as np
import matplotlib.pyplot as plt

def f(x,a,m):
        return a*(x**m)
def df(x,a,m):
        return m*a*(x**(m-1))
def sdc(x,a,m,c1,xk,f,df):
       return f(xk,a,m) + c1 * (x - xk) * df(xk,a,m)

xk = 1
c1 = 1/3
a = 2

X_lim = 1.5
X_curve = np.linspace(-X_lim, X_lim, 100)
X_line = np.array([-X_lim, 1])

plt.figure(figsize=(10, 5))

for m, col1, col2 in [(2, "Skyblue", "orange"), (4, "blue", "red")]:
        plt.plot(xk, f(xk,a,m), 
                label=f"f(1) = {a} * ${xk}^{m}$", 
                color="black", 
                marker='o')
        plt.plot(X_curve, f(X_curve,a,m), 
                label=f"f(x) = {a} * $x^{m}$", 
                color=col1, 
                linestyle="-")  
        plt.plot(X_line, sdc(X_line,a,m,c1,xk,f,df), 
                label=f"Sufficient decrease condition for m = {m}", 
                color=col2, 
                linestyle="--")        
plt.legend(loc='upper center')#, fontsize=12)    
plt.title(f"Sufficient decrease condition for f(x) = a * $x^m$ with a = {a}, and $c_1$ = {c1:.2f}")#, fontsize=24)
plt.xlabel("x")#, fontsize=20)
plt.ylabel("f(x)")#, fontsize=20)
plt.grid(True, which='major', linestyle='--', alpha=0.7)
plt.show()
```
